```python
import os
import csv
from pathlib import Path

def remove_first_row_from_csvs(folder_path):
    # 遍历文件夹内所有CSV文件
    folder = Path(folder_path)
    for csv_file in folder.glob("*.csv"):
        # 读取文件内容并删除首行
        with open(csv_file, 'r', newline='', encoding='utf-8') as f:
            reader = csv.reader(f)
            rows = list(reader)
            if len(rows) < 1:  # 空文件跳过
                continue
            processed_rows = rows[1:]  # 删除第一行
        
        # 覆盖写入原文件
        with open(csv_file, 'w', newline='', encoding='utf-8') as f:
            writer = csv.writer(f)
            writer.writerows(processed_rows)


target_folder = "/home/rwkv/RWKV-TS/gift-eval-datasets/electricityloaddiagrams20112014/LD2011_2014_prcessed"  # 替换为你的文件夹路径
remove_first_row_from_csvs(target_folder)
print("所有CSV文件的首行已删除并保存！")
```
